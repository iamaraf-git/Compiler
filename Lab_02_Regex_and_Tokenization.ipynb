```python
from collections import OrderedDict, defaultdict
# Orderdict= dictionary subclass
#defaultdict = dictionary subclass
def token(text):
  low=text.lower()
  print("lower case: ",low)

#Find tokens
  tok=re.findall(r'\w+|[^\w\s]',low)
  print("Tokens are: ", tok)

#count frequency through dictionary class named defaultdict
  freq= defaultdict(int)
  for t in tok:
    freq[t]+=1

#maintain order, remove duplicacy of words through dictionary class named OrderedDict
  order_freq= OrderedDict()
  for t in tok:
    if t not in order_freq:
      order_freq[t]=freq[t]

  return tok, order_freq

text= "Welcome to Compiler Construction lab course.\
Here you can learn NLP through compiler construction."

#word for tok and frequency for order_freq
word,frequency=token(text)
count=len(word)
print(f"There are {count} tokens in the text")

#show word frequencies through dictionary key-value pair
print("Frequency are: ")
for key, value in frequency.items():
  print(f"{key:} {value}")



```
