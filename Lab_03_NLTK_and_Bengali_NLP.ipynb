```python
from nltk import word_tokenize, sent_tokenize
# word tokenization
sent= "Welcome to Compiler Construction lab course. Here you can learn NLP through compiler construction."
print("Word tokenized: ", word_tokenize(sent))
print("Tokens are: ", len(word_tokenize(sent)))
# senetence tokenization
sent1=sent_tokenize(sent)
print("Sentence tokenized: ", sent1)
print("Tokens are: ", len(sent_tokenize(sent)))
print("sentence value in index: ",sent1[0])
```

```python
from nltk.probability import FreqDist
# probability= is a module name
#NLTK= is a library
#FreqDist= is a class

sent= "Welcome to Compiler Construction lab1 course. Here you can learn NLP through compiler construction."
print(sent)

#lower each words
sentlow=sent.lower()
print("lower case: ",sentlow)

# sentence tokenization
senttoken=sent_tokenize(sentlow)
print("Sentence tokenized: ", senttoken)

# word tokenization
token=word_tokenize(sentlow)
print("Word tokenized: ", token)

#frequency count
freqdist1= FreqDist(token)
print("Frequency are: ")
for key, value in freqdist1.items():
  print(f"{key:} {value}")



```

```python
from indicnlp.tokenize import indic_tokenize
from indicnlp.tokenize import sentence_tokenize

bsent="আমি বাংলা তে লিখছি । তুমিও কি বাংলা তে লিখছো?"

#apply indic tokenize for word tokenization
ibtoken=indic_tokenize.trivial_tokenize(bsent)
print("bengali word: ", ibtoken)

#apply sentence_tokenize for sentence tokenization
ibsent=sentence_tokenize.sentence_split(bsent, lang='bn')
print("bengali senetence: ", ibsent)
print(ibsent[1])
```
